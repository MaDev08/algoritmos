```python
#NÍVEL 4: 

soma = 0 
quantidade = 0 
  

valor = float(input('Digite um valor positivo ou -1 para ' 
                     'sair: ')) 
  
while valor != -1: 
    soma += valor    
    quantidade += 1      
    valor = float(input('Digite um valor positivo ou -1 '  'para sair: ')) 
  

if quantidade > 0: 
    media = soma / quantidade 
    print(f'Soma:       {soma:.2f}') 
    print(f'Quantidade: {quantidade}') 
    print(f'Media:      {media:.2f}') 
else: 
    print('Nenhum valor foi digitado.')


```
