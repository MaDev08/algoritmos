```python
#Nível4

"""
Situação: estudar inglês

Decomposição (como vou estudar inglês?)

1-) Lê um texto em inglês
2-) Lê a tradução em português
3-) Anota as palavras novas que não sabe
4-) Anota a tradução das palavras novas que não sabe
5-) Adiciona todas as palavras novas com tradução no app AnkiDroid Flashcards
6-) Estude as palavras novas
7-) Revisa as palavras novas aprendidas

Entrada: lê texto em inglês / traduz em português
Saída: revisar as palavras novas / aprender novas palavras
"""

"""
ALGORITMO:

1-) Digitar ou colar o texto em inglês
2-) Ler a tradução em português
3-) Anotar todas as palavras novas que não conhece
4-) Anotar a tradução de todas as palavras novas que não conhece
5-) Adicionar todas as palavras novas no Ankidroid
6-) Estudar todas as palavras novas no Ankidroid
7-) Revisão das palavras aprendidas no Ankidroid

"""


#CÓDIGO GERADO COM IA A PARTIR DO FLUXOGRAMA:

def estudar_ingles():
    # 1. Entrada do texto em inglês
    print("--- PASSO 1: Ler um texto em inglês ---")
    texto_ingles = input("Digite ou cole o texto em inglês: ")

    # 2. Entrada da tradução em português
    print("\n--- PASSO 2: Ler a tradução em português ---")
    traducao_texto = input("Digite ou cole a tradução em português: ")

    # 3 e 4. Anotação de palavras novas e suas traduções
    print("\n--- PASSO 3 e 4: Anotar palavras novas e suas traduções ---")
    palavras_novas = []

    while True:
        palavra = input("Digite uma palavra nova que não conhece (ou 'fim' para encerrar): ").strip()
        if palavra.lower() == 'fim':
            break
        if palavra == '':
            continue

        traducao_palavra = input(f"Digite a tradução de '{palavra}': ").strip()
        palavras_novas.append({
            "ingles": palavra,
            "portugues": traducao_palavra
        })

    if not palavras_novas:
        print("\nNenhuma palavra nova foi anotada. Bons estudos!")
        return

    # 5. Adicionar ao AnkiDroid (Simulação)
    print("\n--- PASSO 5: Adicionando palavras ao AnkiDroid (Simulação) ---")
    for item in palavras_novas:
        print(f"[AnkiDroid] Cartão adicionado: Frente: {item['ingles']} | Verso: {item['portugues']}")

    # 6. Estudar as palavras
    print("\n--- PASSO 6: Estudar as palavras novas ---")
    for item in palavras_novas:
        print(f"Estudando: {item['ingles']} significa {item['portugues']}")

    # 7. Revisar as palavras aprendidas (Mini-quiz)
    print("\n--- PASSO 7: Revisão das palavras aprendidas ---")
    acertos = 0
    for item in palavras_novas:
        resposta = input(f"Qual é a tradução de '{item['ingles']}'? ").strip()
        if resposta.lower() == item['portugues'].lower():
            print("✅ Correto!")
            acertos += 1
        else:
            print(f"❌ Incorreto. A tradução correta é '{item['portugues']}'.")

    # Saída e Resultados
    print("\n=== RESULTADO DA SESSÃO ===")
    print("✔ Texto em inglês e tradução lidos.")
    print(f"✔ Você aprendeu e revisou {len(palavras_novas)} novas palavras.")
    print(f"✔ Desempenho na revisão: {acertos} de {len(palavras_novas)} acertos ({ (acertos/len(palavras_novas))*100 :.1f}%).")

# Executar o algoritmo
estudar_ingles()
```
