```python
#NÍVEL 2:

compra = float(input('Qual o valor da sua compra? \n'))

if compra > 200: 
    desconto = compra * 0.10 
    valor_final = compra - desconto 
    print(f'O valor com desconto foi de {valor_final}')
else: 
    print(f'Sua compra não teve desconto')
```

```python
#NÍVEL 4:

p1 = float(input('Qual é a nota da prova 1? (PESO DE 30%)'))
p2 = float(input('Qual é a nota da prova 2? (PESO DE 20%)')) 
p3 = float(input('Qual é a nota da prova 3? (PESO DE 10%)'))
proj = float(input("Nota Projetos (peso 20%): ")) 

media = (p1 * 0.30) + (p2 * 0.20) + (p3 * 0.10) + (proj * 0.20) 

if media >= 7: 
    print('aprovado')
elif media >= 5: 
    print('recuperação')
else: 
    print('reprovado')
```
